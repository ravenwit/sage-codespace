```python
A = matrix(QQ, [[0, 1], [1, 0]]) # Pauli-X over rationals
B = matrix(QQ, [[1, 0], [0, -1]]) # Pauli-Z over rationals
commutator = A * B - B * A
print("Commutator [X, Z]:\n", commutator)
print("Eigenvalues of X + Z:", (A + B).eigenvalues())
```

```python
u, v = var("u v")
# Torus parameterization
R, r = 2.0, 0.7
px = (R + r * cos(v)) * cos(u)
py = (R + r * cos(v)) * sin(u)
pz = r * sin(v)
parametric_plot3d((px, py, pz), (u, 0, 2*pi), (v, 0, 2*pi), aspect_ratio=1)
```
